# Arrays
## .arange, .linspace, .delete, .fullspace

```python
# Array mit pre Werten erstellen
# np.arange(start,end,stepsize)
# end non-inclusive
np.arange(2.0,3.0,0.25)
```

```python
# np.linspace(start,stop,num,endpoint=True)
# stop non-inclusive
np.linspace(0,1,11,endpoint=False)
```

```python
# array=np.delete(array,index(-array))
# Bsp.
U=[1,2,3,4]
np.delete(U,[0,-1])
```

```python
#slicing np.s_
U=np.arange(1,10,1)
np.delete(U, np.s_[-3:])
```

```python
# aus einem wert ein array mit n-mal dem gleichen wert machen
A=np.array([1,2,3,4,5])
a=1
np.full_like(A,a)
```

## .vstack

```python
# The array formed by stacking the given arrays, will be at least 2-D.
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])
np.vstack((a,b))
```

## .concatenate

```python
# Join a sequence of arrays along an existing axis.
np.concatenate((a,b))
```

## .sort

```python
# sorting
numbers=np.array([40,30,20,10,6,3,0,-3,-6,-10,-20,-30,-40])
numbers=np.sort(numbers)
print(numbers)
# .join zum Einfügen eines Strings zwischen allen array werten
text =" & ".join(numbers.astype(str))
print(text)
```

## .argmin

```python
# Wert im Array finden
zielwert = 6
numbers=np.array([40,30,20,10,6,3,0,-3,-6,-10,-20,-30,-40])
index=np.abs(numbers-zielwert).argmin()
print(index)
```
