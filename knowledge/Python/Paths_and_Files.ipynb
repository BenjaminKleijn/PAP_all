## DataWriter und Path_initialize class


```python
class Path_initialize():
    def __init__(self):
        self.cwd_path=Path.cwd()

    def set_Path(self,new_path):
        self.new_path=Path.cwd()/new_path
        print("New path is setted:",self.new_path)

    def check_Path(self):
        print("This is your current Path",self.new_path)

class Data_Writer(Path_initialize):

    def set_file(self,filename):
        self.file_path=Path(self.new_path/filename)

    def check_file(self):
        if self.file_path.exists():
            print("This file exists")
        else:
            print("This file was not found.")

    def read_file(self,filename=None,default_path=None):
        if filename:
            self.set_file(filename)
        with open(self.file_path,"r",encoding="utf-8") as f:
            print(f.read())

    def clear_file(self,filename=None):
        if filename:
            self.set_file(filename)
        with open(self.file_path,"w",encoding="utf-8") as f:
            pass
        # print(self.read_file())
```
