```python
# Übergebe Liste an Zeichen, die entfernt werden sollen, und Dictionary an Dinge die replaced werden
def convert_symbols(valstr, remove=None, replace=None):         
    if remove:
        for char in remove:
            valstr = valstr.replace(char, "")
            
    if replace:
        for old, new in replace.items():
            valstr = valstr.replace(old, new)

    return (valstr)
convert_symbols(data,remove=["\n","R"],replace={" ":"1"})
```

# Delete Vocals

```python
def deletevocals(text):
    # Einrückungen entfernen (gemeinsame Tabs des gesamten Textes)
    text = textwrap.dedent(text)
    #  entfernt Bindestriche: - findet Bindestrich, \n Zeilenumbruch direkt dahinter, \s Whitespaces *beliebig viele 
    text = re.sub(r'-\n\s*',"",text)
    # entfernt Zeilenumbrüche
    text = text.replace("\n"," ")
    
    # Eine Funktion, die vom re.sub aufgerufen wird, wenn ein Treffer erzielt wird
    def convert(match):
        # match.group(1) ist der Buchstabe nach dem Pünktchen (z.B. 'a' oder 'A')
        buchstabe = match.group(1)
        # Dictionary für die echten Umlaute
        umlaute = {'a': 'ä', 'u': 'ü', 'o': 'ö', 'A': 'Ä', 'U': 'Ü', 'O': 'Ö'}
        # Gibt den Umlaut zurück. Falls das Zeichen nicht im Dict ist, lass es wie es ist, denn .get(key,default)
        return umlaute.get(buchstabe, buchstabe)
    # Das Pattern sucht nach ¨ gefolgt von einem beliebigen Buchstaben aus der Auswahl, [X] findet genau ein Zeichen aus Menge X 
    # und übergibt das 1.Zeichen aus der Gruppe, den Buchstaben, an die convert Funktion 
    return re.sub(r'¨([auoAUO])', convert, text) 
```

```python
# Der König hat erklärt, wie man das automatisch für alle files in einem ordner macht, also einfach ordner in vs code öffnen oder rechtsclick auf ordner und dann in terminal öffnen
Get-ChildItem -Recurse -Filter *.tex | ForEach-Object {
    (Get-Content $_.FullName -Raw -Encoding Default) -replace '(?<!\\)\$([^$]+)(?<!\\)\$', '\($1\)' | Set-Content $_.FullName -Encoding Default
}
# es wird gecheckt ob davor ein \$ steht, also eine Geldangabe, in powershell wird auf eine gruppe mit $1 zurpckgegriffen
```

```python
# für einfaches ersetzen von Text ohne regex
Get-ChildItem -Recurse -Filter main.tex | ForEach-Object {
    (Get-Content $_.FullName -Raw -Encoding utf8).Replace('\subsection', '\section') | Set-Content $_.FullName -Encoding utf8
}
```
