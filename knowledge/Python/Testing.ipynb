```python
# mehrere subplots:
# dictionary
axesinformation={'key1':'info','key2':'info','key3':'info','key4':'info'}
# 4 subplots
fig, axs = plt.subplots(2, 2, figsize=figsize_cm(4,4))
# holt sich alle objekte (axs =[[ax1, ax2], [ax3, ax4]] muss erstmal zu axs.flatten=[ax1,ax2,ax3,ax4] gemacht werden)
for ax, (key, info) in zip(axs.flatten(), axesinformation.items()):
    ax.set_title(f'{key}')
    ax.scatter(1,1)
plt.tight_layout()
# plt.savefig(Ausgabe_path / 'klines.png', format="png", bbox_inches='tight')
plt.show()
```

```python
# ('info1','info2') wird als array ausgegeben, wenn man schreibt ax(key,info), dann müsste man erst info=[info1,info2] entpacken. 
# sets mit {} haben keine reihenfolge, also wenn man nur eine liste von einzelnen infos hat, dann [] für array/liste nutzen oder () für ein Tupel(hat auch order)
dictionary={'key1':('1','info2'),'key2':('2','info2')}
fig, axs = plt.subplots(2, 2, figsize=figsize_cm(4,4))
for ax, (key,(info1,info2)) in zip(axs.flatten(),dictionary.items()):
    ax.set_title(f'{info1}')
    ax.scatter(1,1)
plt.tight_layout()
# plt.savefig(Ausgabe_path / 'klines.png', format="png", bbox_inches='tight')
plt.show()
# zwei letzten plots kriegen nichts, weil zip. denen keine items aus dictionary holen kann und deswegen abbricht. 
```

## lambda functions

```python
# out of multiple arg function ff build single arg f 
def ff(arg1,arg2):
    return arg1+arg2
f = lambda arg1, arg2=1: ff(arg1, arg2)
print(f(1))
```
